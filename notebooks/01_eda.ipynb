```python
import sys
from pathlib import Path

# Add project root to Python path
ROOT = Path.cwd().parent
sys.path.append(str(ROOT))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from src.data.data_loader import download_stock_data
```

```python
df = download_stock_data("NVDA", "5y")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDate Range:")
print("Start:", df.index.min())
print("End:", df.index.max())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Dates:")
print(df.index.duplicated().sum())
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df.index, df["Close"])

plt.title("NVDA Closing Price - 5 Years")
plt.xlabel("Date")
plt.ylabel("Price")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df.index, df["Volume"])

plt.title("NVDA Trading Volume - 5 Years")
plt.xlabel("Date")
plt.ylabel("Volume")

plt.grid(True)
plt.show()
```

```python
df["DailyReturn"] = df["Close"].pct_change()

df[["Close", "DailyReturn"]].head(10)
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df["DailyReturn"].dropna(), bins=50)

plt.title("NVDA Daily Return Distribution")
plt.xlabel("Daily Return")
plt.ylabel("Frequency")

plt.grid(True)
plt.show()
```

```python
print("Daily Return Statistics:")
print(df["DailyReturn"].describe())
```

```python
df["Volatility20"] = df["DailyReturn"].rolling(20).std()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df.index, df["Volatility20"])

plt.title("NVDA 20-Day Rolling Volatility")
plt.xlabel("Date")
plt.ylabel("Volatility")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df.index, df["Open"], label="Open")
plt.plot(df.index, df["High"], label="High")
plt.plot(df.index, df["Low"], label="Low")
plt.plot(df.index, df["Close"], label="Close")

plt.title("NVDA OHLC Prices")
plt.xlabel("Date")
plt.ylabel("Price")

plt.legend()
plt.grid(True)
plt.show()
```
